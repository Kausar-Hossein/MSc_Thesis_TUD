```python
# =========================================================
# KIRCHNER-STYLE VALIDATION
# Calibration: 2010 only
# Validation: 2011–2022
# Station ID, name, and area are added manually in the code
# One recession plot per station
# =========================================================

import os
import glob
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# =========================
# SETTINGS
# =========================

discharge_folder = "/kaggle/input/datasets/kausar15027/gw-discharge-data/discharge/03_all"

output_folder = "/kaggle/working/recession_2010_calibration_validation_manual_station_list"
os.makedirs(output_folder, exist_ok=True)

start_year = 2010
end_year = 2022
calibration_year = 2010
validation_start = 2011
validation_end = 2022

min_recession_length = 7

# =========================
# MANUAL STATION LIST
# Format:
# Station_ID    NAME    Area_km2
# =========================

station_data_text = r"""
560051	Bad D??ben 1	6159.23
560021	Golzern 1	5428.98
560000	K??ssern	5352.45
566100	Erlln	2978.18
566085	Leisnig	2875.01
562140	Grosssermuth	2350.42
562135	Colditz	2318.06
562115	Wechselburg 1	2097.21
582820	Spreewitz	2078.55
582080	Sprey	1645.25
567451	Lichtenwalde 1	1570.67
562090	Wolkenburg	1415.28
562070	Zwickau-P??lbitz	1019.94
577571	B??hlen 1	969.217
576440	Elsterberg	944.882
566055	Mahlitzsch	914.627
568185	Fl??ha 1	793.427
582060	Lieske	775.84
568175	Hetzdorf 1	754.504
562040	Niederschlema	753.245
567438	Fl??ha 2	729.917
554560	Zabeltitz 1	724.837
554550	Kleinrasch??tz	679.076
567435	Kunnersdorf	677.536
562031	Aue 3	679.794
583200	Boxberg	658.726
576430	Plauen	657.134
553020	Neuwiese	652.259
568160	Borstendorf	638.902
576421	Stra?ƒberg	611.804
583223	Reichwalde 3	596.095
566040	Nossen 1	586.01
566030	Nossen 2	565.795
564410	G??ritzhain	532.242
567420	Hopfgarten	529.809
568150	Pockau 2	514.773
564400	Chemnitz 1	402.313
568140	Pockau 1	385.311
551208	Friedrichstadt	379.977
551200	Doelzschen	371.32
563790	Aue 1	362.707
566025	Kleinvoigtsberg	353.439
582052	Spreewiese OP	333.042
551190	Hainsberg 6	328.089
576411	Oelsnitz 1	325.084
583302	Hammerstadt 1	324.214
563780	Schwarzenberg	316.39
568133	Olbernhau 3	314.168
578110	Leipzig-Thekla	314.261
552411	Torgau-Sued 1	307.161
554520	Gro?ƒdittmannsdorf	298.708
567320	Niederstriegis 1	285.121
582035	Bautzen 1	277.705
550190	Porschdorf 1	266.917
583350	Burgneudorf	258.679
567310	B??hrigen	252.532
577911	Borna 1	240.842
550810	Elbersdorf	227.18
552210	Merzdorf	209.502
567850	Streckewalde	206.076
583122	Gr??ditz 2	200.506
562014	Neidhardtsthal 1	197.272
550940	Dohna	193.37
552305	Klingenhain	189.849
554220	Zescha	180.566
577901	Streitwald 1	179.297
582010	Schirgiswalde	178.398
583271	Sproitz	175.277
550931	M??hlbach 1	172.088
576401	Adorf 1	171.359
663090	Tauchritz	166.47
563950	Wildenau 1	166.368
577501	Neukirchen 1	165.358
551820	Garsebach	164.19
551351	Hainsberg 7	162.276
662011	Gro?ƒsch??nau 2	162.184
553013	Trado 3	160.413
577220	Mylau	157.754
550110	Kirnitzschtal	154.05
552110	Seerhausen 1	153.906
583352	Burg 2	151.53
552012	Piskowitz 2	150.732
562012	Sch??nheide 3	148.305
564531	Altchemnitz 2	143.457
578091	Albrechtshain 1	137.318
552401	Melpitz 1	137.235
564620	Harthau	136.015
583290	S?ñrichen	134.273
561000	Thallwitz	133.512
568121	Oberneusch??nberg	133.242
550620	Neundorf	132.649
567200	Naundorf	130.834
567221	Krummenhennersdorf 1	130.834
583251	J?ñnkendorf 1	124.502
568400	Z??blitz	123.935
564290	St. Egidien	122.288
564120	Wilkau	117.121
551811	Munzig 1	114.505
554510	Radeberg	112.574
560950	Kleinzschepa	106.502
550091	Buschm??hle 1	106.144
583345	Litschen	105.965
577470	Grossstolpen	105.629
554120	Sch??nau	104.417
554210	Prischwitz	103.969
564611	Jahnsdorf 1	103.261
550302	Sebnitz 2	101.769
568105	Neuhausen 1	101.159
550501	Koenigstein 3	100.71
577100	Hasenm??hle	100.073
550290	Sebnitz 3	98.7326
554605	M??hlbach	97.543
577490	Werdau	96.4098
564505	Burkhardtsdorf 2	93.0147
567400	Tannenberg	91.098
554411	Reichenau 1	88.111
562010	Rautenkranz	87.6523
567700	Wiesa	86.173
567194	Niederbobritzsch	83.7
583340	Lippitsch	78.5325
663000	Rennersdorf 3	78.2399
551420	Dippoldiswalde 1	77.3407
565990	Mulda 1	76.8718
568102	Rauschenbach 3	76.7035
568350	Rothenthal	76.1312
577211	Rodewisch 1	75.7391
662001	Seifhennersdorf 1	75.2113
551005	Lockwitz	71.6065
554570	Radeburg 3	70.4061
550913	Lauenstein 4	69.2577
550800	Bischofswerda	68.85
577068	Stra?ƒberg 2	66.3878
567964	Schl???ƒchen	65.1897
552101	Ostrau 1	63.4015
567840	Schmalzgrube 2	62.3341
552190	Nebitzschen	59.2193
582005	Neusalza-Spremberg	58.8712
568080	Deutschgeorgenthal 1	58.1299
577880	Langenleuba-Niederhain	58.0774
560301	Nemt 1	56.8095
530020	Klingenthal 1	55.5419
568880	Hammerleubsdorf	55.4609
583280	Holtendorf	54.3928
551411	Schmiedeberg 1	53.6236
554565	Kleinwolmsdorf	53.0247
567301	Obersch??na 1	52.7683
583020	Hohberg	50.5984
564201	Niederm??lsen 1	49.3679
551310	Ammelsdorf	49.334
553001	Kamenz 1	49.2293
567590	Annaberg 1	48.7269
550710	Markersbach	48.0308
576391	Bad Elster 1	47.6998
564100	Kirchberg	46.8559
551802	Herzogswalde 2	46.2714
560201	Neichen 1	46.2207
552000	Ziegenhain	46.1699
564346	Uhlsdorf	46.12
577026	G??rnitz	45.6409
551000	Kreischa	43.8062
583235	Sch??ps	42.5107
554260	Pietzschwitz	42.2443
583090	Gro?ƒschweidnitz	41.4093
554665	Stroga	41.2401
567890	J??hstadt 4	40.6996
550390	Neustadt 1	40.1121
551100	Klotzsche	39.5266
567356	Hainichen	39.4943
565980	Bienenm??hle	38.8889
567120	Lichtenberg 2	38.8781
567000	Wolfsgrund	37.8055
550490	Bielatal 1	37.6058
563890	Markersbach 2	37.1316
567390	Schlettau	36.023
550640	Gottleuba 3	35.4653
577000	Sohl	35.3732
577870	Langenleuba-Niederhain	35.042
554100	Panschwitz	34.3477
577455	Wildenhain	33.4923
583230	Oehlisch	31.4395
564500	Niederzw??nitz	31.3749
550560	Cunnersdorf 1	30.9079
578060	Gro?ƒbardau 1	30.4341
563880	Markersbach 1	29.7973
567166	Freiberg	28.9324
583170	Kotitz	28.9007
662096	Niederoderwitz	28.8023
577028	Untertriebelbach	28.1785
551601	Wilsdruff 1	26.9388
583342	Lomske	26.6582
550901	Lauenstein ZP	26.6184
563800	Johanngeorgenstadt 2	26.0916
550762	Liebstadt 2	25.269
577040	Ramoldsreuth	23.6271
568600	Saidenbach ZP 1	22.2833
564160	Wolfersgr??n	21.6973
552143	Paltzschen 3	20.4117
550961	Geising 1	20.0424
564736	R??hrsdorf	19.9921
577120	Neuensalz	19.8542
582000	Ebersbach	16.0724
551302	Rehefeld 2	15.4242
551570	Potschappel	13.6888
663131	Rennersdorf 6	13.4864
551521	Reichst?ñdt	12.6974
551950	Niederau	12.0096
551561	Freital 1	11.9908
577030	Bobenneukirchen	11.8917
577700	Langenreinsdorf	10.4023
577070	Werda ZP	9.55927
577010	Sch??nlind 2	9.40037
568200	Deutschgeorgenthal 2	8.73116
551590	Gorbitz 2	7.39763
568300	Rauschenbach 2	7.28803
583190	Drehsa	7.27249
567650	Cranzahl	7.16053
578170	Gro?ƒbardau	7.04174
568250	Neuwernsdorf	6.91631
563290	Sachsengrund	6.76033
551510	B?ñrenfels	6.26949
563550	Sosa ZP 2	4.81614
568310	Caemmerswalde	4.14114
578050	Glasten	3.8738
551580	Gorbitz 1	3.39808
583140	Wuischke	2.90904
550650	Gottleuba 2	2.38556
577025	Schoeneck	1.95587
563401	Carlsfeld ZP 1	1.69298
564660	Stollberg ZP	0.902293
662021	Zittau 5	295.702
550630	Gottleuba 1	29.7705
583282	Koenigshain	7.65023
568700	Neunzehnhain II ZP 1	10.3183
564300	Niederlungwitz	137.705
663100	Rennersdorf 1	62.7456
664100	Krauschwitz	85.9718
568740	Neunzehnhain II ZP 2	0.995891
231	Wernersbach HP	4.60601
662022	Zittau 6	277.544
566010	Berthelsdorf	244.399
"""

# =========================
# FUNCTIONS
# =========================

def load_station_list_from_text(station_data_text):
    """
    Reads station ID, station name, and area directly from pasted text.

    Rule:
    - first value  = station_id
    - last value   = area_km2
    - middle text  = station_name
    """
    rows = []

    for line in station_data_text.strip().splitlines():
        line = line.strip()

        if not line:
            continue

        parts = line.split()

        if len(parts) < 3:
            continue

        try:
            station_id = int(parts[0])
            area_km2 = float(parts[-1])
            station_name = " ".join(parts[1:-1])

            rows.append({
                "station_id": station_id,
                "station_name": station_name,
                "area_km2": area_km2
            })

        except Exception:
            continue

    station_df = pd.DataFrame(rows)

    if station_df.empty:
        raise ValueError("Manual station list could not be read. Please check the station_data_text format.")

    station_df = station_df.drop_duplicates(subset=["station_id"]).reset_index(drop=True)

    return station_df


def read_table_flexible(file_path):
    """
    Reads discharge CSV files flexibly.
    Handles encoding and separator problems.
    """
    encodings = ["utf-8", "utf-8-sig", "latin1", "cp1252"]
    separators = [",", ";", "\t"]

    for enc in encodings:
        for sep in separators:
            try:
                df = pd.read_csv(file_path, encoding=enc, sep=sep)
                if df.shape[1] >= 2:
                    return df
            except Exception:
                pass

    return pd.read_csv(file_path, encoding="latin1")


def power_law(Q, a, b):
    return a * Q**b


def detect_date_column(df):
    cols = [
        c for c in df.columns
        if "date" in str(c).lower()
        or "time" in str(c).lower()
        or "datum" in str(c).lower()
    ]

    if len(cols) == 0:
        raise ValueError("No date/time column found.")

    return cols[0]


def detect_discharge_column(df):
    possible = [
        c for c in df.columns
        if "discharge" in str(c).lower()
        or str(c).lower().strip() == "q"
        or "flow" in str(c).lower()
        or "value" in str(c).lower()
        or "abfluss" in str(c).lower()
    ]

    if len(possible) > 0:
        return possible[0]

    numeric_cols = df.select_dtypes(include=[np.number]).columns

    if len(numeric_cols) == 0:
        raise ValueError("No numeric discharge column found.")

    return numeric_cols[-1]


def extract_station_id_from_filename(file):
    fname = os.path.basename(file)

    match = re.match(r"(\d+)", fname)

    if match:
        return int(match.group(1))

    nums = re.findall(r"\d+", fname)

    if len(nums) > 0:
        return int(nums[0])

    return None


def detect_recession_periods(df, q_col, min_len=7):
    """
    Detect continuous recession periods where Q decreases day by day.
    Keeps recession group ID to avoid shifting across separate recession events.
    """
    df = df.copy()
    df["dQ"] = df[q_col].diff()
    df["is_recession"] = df["dQ"] < 0

    recession_frames = []
    group_id = 0
    current_indices = []

    for idx, is_rec in df["is_recession"].items():
        if is_rec:
            current_indices.append(idx)
        else:
            if len(current_indices) >= min_len:
                temp = df.loc[current_indices].copy()
                temp["recession_group"] = group_id
                recession_frames.append(temp)
                group_id += 1

            current_indices = []

    if len(current_indices) >= min_len:
        temp = df.loc[current_indices].copy()
        temp["recession_group"] = group_id
        recession_frames.append(temp)

    if len(recession_frames) == 0:
        return pd.DataFrame()

    return pd.concat(recession_frames).sort_index()


def prepare_recession_data(file):
    df = read_table_flexible(file)
    df.columns = df.columns.astype(str).str.strip()

    date_col = detect_date_column(df)
    q_col = detect_discharge_column(df)

    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")
    df[q_col] = pd.to_numeric(df[q_col], errors="coerce")

    df = df[[date_col, q_col]].dropna()
    df = df.set_index(date_col).sort_index()

    df = df.loc[f"{start_year}-01-01":f"{end_year}-12-31"]

    if df.empty:
        return pd.DataFrame()

    df = df.resample("D").mean().dropna()

    if len(df) < 50:
        return pd.DataFrame()

    rec = detect_recession_periods(df, q_col, min_recession_length)

    if rec.empty:
        return pd.DataFrame()

    # Shift only inside each recession group
    rec["Q_next"] = rec.groupby("recession_group")[q_col].shift(-1)

    rec["Q_avg"] = (rec[q_col] + rec["Q_next"]) / 2.0
    rec["minus_dQdt"] = -(rec["Q_next"] - rec[q_col])

    rec = rec.replace([np.inf, -np.inf], np.nan).dropna()

    rec = rec[
        (rec["Q_avg"] > 0)
        & (rec["minus_dQdt"] > 0)
    ].copy()

    rec["year"] = rec.index.year

    return rec


def nse(obs, sim):
    obs = np.array(obs, dtype=float)
    sim = np.array(sim, dtype=float)

    mask = np.isfinite(obs) & np.isfinite(sim)
    obs = obs[mask]
    sim = sim[mask]

    if len(obs) < 5:
        return np.nan

    denominator = np.sum((obs - np.mean(obs)) ** 2)

    if denominator == 0:
        return np.nan

    return 1 - np.sum((obs - sim) ** 2) / denominator


def rmse(obs, sim):
    obs = np.array(obs, dtype=float)
    sim = np.array(sim, dtype=float)

    mask = np.isfinite(obs) & np.isfinite(sim)
    obs = obs[mask]
    sim = sim[mask]

    if len(obs) == 0:
        return np.nan

    return np.sqrt(np.mean((obs - sim) ** 2))


# =========================
# LOAD MANUAL STATION LIST
# =========================

station_df = load_station_list_from_text(station_data_text)

print("Number of manually added stations:", len(station_df))
display(station_df.head())

station_ids = set(station_df["station_id"].astype(int).tolist())

# =========================
# FIND AVAILABLE DISCHARGE FILES
# =========================

csv_files = glob.glob(os.path.join(discharge_folder, "*.csv"))

file_records = []

for file in csv_files:
    sid = extract_station_id_from_filename(file)

    if sid is None:
        continue

    if sid in station_ids:
        file_records.append({
            "station_id": sid,
            "file": file,
            "file_name": os.path.basename(file)
        })

available_files_df = pd.DataFrame(file_records)

if available_files_df.empty:
    raise ValueError("No matching discharge CSV files found for the manually added station IDs.")

available_files_df = available_files_df.drop_duplicates(subset=["station_id"]).reset_index(drop=True)

print("Number of matching discharge files found:", len(available_files_df))
display(available_files_df.head())

# Merge station info with file paths
run_df = available_files_df.merge(station_df, on="station_id", how="left")

matched_path = os.path.join(output_folder, "matched_station_files.csv")
run_df.to_csv(matched_path, index=False)

print("Matched station file list saved to:")
print(matched_path)

# =========================
# RUN VALIDATION
# =========================

all_results = []
failed_results = []

for _, row in run_df.iterrows():

    station_id = int(row["station_id"])
    station_name = row.get("station_name", "")
    area_km2 = row.get("area_km2", np.nan)
    file = row["file"]
    fname = row["file_name"]

    print("\nProcessing station:", station_id, "|", station_name)

    try:
        rec = prepare_recession_data(file)

        if rec.empty:
            print("No recession data found.")
            failed_results.append({
                "station_id": station_id,
                "station_name": station_name,
                "area_km2": area_km2,
                "file_name": fname,
                "reason": "No recession data found"
            })
            continue

        calib = rec[rec["year"] == calibration_year].copy()

        valid = rec[
            (rec["year"] >= validation_start)
            & (rec["year"] <= validation_end)
        ].copy()

        if len(calib) < 20 or len(valid) < 20:
            print("Not enough calibration or validation points.")
            failed_results.append({
                "station_id": station_id,
                "station_name": station_name,
                "area_km2": area_km2,
                "file_name": fname,
                "reason": f"Not enough points: calibration={len(calib)}, validation={len(valid)}"
            })
            continue

        # Fit power law using only 2010 calibration data
        popt, _ = curve_fit(
            power_law,
            calib["Q_avg"].values,
            calib["minus_dQdt"].values,
            p0=[0.01, 2.0],
            bounds=([0, 0], [np.inf, 10]),
            maxfev=50000
        )

        a, b = popt

        calib["pred_minus_dQdt"] = power_law(
            calib["Q_avg"].values,
            a,
            b
        )

        valid["pred_minus_dQdt"] = power_law(
            valid["Q_avg"].values,
            a,
            b
        )

        NSE_calibration = nse(
            calib["minus_dQdt"].values,
            calib["pred_minus_dQdt"].values
        )

        RMSE_calibration = rmse(
            calib["minus_dQdt"].values,
            calib["pred_minus_dQdt"].values
        )

        NSE_validation = nse(
            valid["minus_dQdt"].values,
            valid["pred_minus_dQdt"].values
        )

        RMSE_validation = rmse(
            valid["minus_dQdt"].values,
            valid["pred_minus_dQdt"].values
        )

        all_results.append({
            "station_id": station_id,
            "station_name": station_name,
            "area_km2": area_km2,
            "file_name": fname,
            "calibration_year": calibration_year,
            "validation_years": f"{validation_start}-{validation_end}",
            "a": a,
            "b": b,
            "NSE_calibration": NSE_calibration,
            "RMSE_calibration": RMSE_calibration,
            "NSE_validation": NSE_validation,
            "RMSE_validation": RMSE_validation,
            "n_calibration_points": len(calib),
            "n_validation_points": len(valid),
            "n_total_recession_points": len(rec),
            "min_recession_length_days": min_recession_length
        })

        # =========================
        # SAVE STATION RECESSION DATA
        # =========================

        station_rec_path = os.path.join(
            output_folder,
            f"recession_data_station_{station_id}.csv"
        )

        rec_out = rec.copy()
        rec_out["station_id"] = station_id
        rec_out["station_name"] = station_name
        rec_out["area_km2"] = area_km2
        rec_out.to_csv(station_rec_path)

        # =========================
        # PLOT
        # =========================

        q_min = rec["Q_avg"].min()
        q_max = rec["Q_avg"].max()

        if q_min <= 0 or q_max <= 0 or not np.isfinite(q_min) or not np.isfinite(q_max):
            print("Invalid Q range for plotting.")
            continue

        Q_plot = np.logspace(
            np.log10(q_min),
            np.log10(q_max),
            200
        )

        plt.figure(figsize=(8, 6))

        plt.scatter(
            valid["Q_avg"],
            valid["minus_dQdt"],
            s=8,
            alpha=0.25,
            label=f"Validation points: {validation_start}–{validation_end}"
        )

        plt.scatter(
            calib["Q_avg"],
            calib["minus_dQdt"],
            s=16,
            alpha=0.80,
            label=f"Calibration points: {calibration_year}"
        )

        plt.plot(
            Q_plot,
            power_law(Q_plot, a, b),
            linewidth=2.5,
            label=f"2010 fit: -dQ/dt = {a:.4g} Q$^{{{b:.2f}}}$"
        )

        plt.xscale("log")
        plt.yscale("log")

        plt.xlabel("Discharge, Q")
        plt.ylabel("Recession rate, -dQ/dt")

        plt.title(
            f"Station {station_id} - {station_name}\n"
            f"Calibration: {calibration_year} | Validation NSE = {NSE_validation:.2f}"
        )

        plt.legend()
        plt.grid(True, which="both", linestyle="--", alpha=0.4)

        plot_path = os.path.join(
            output_folder,
            f"recession_station_{station_id}_calib_2010_valid_2011_2022.png"
        )

        plt.savefig(plot_path, dpi=300, bbox_inches="tight")
        plt.show()

        print(
            f"Station {station_id}: "
            f"a={a:.6f}, b={b:.3f}, "
            f"NSE_val={NSE_validation:.3f}, RMSE_val={RMSE_validation:.6f}"
        )

    except Exception as e:
        print(f"Failed station {station_id}: {e}")

        failed_results.append({
            "station_id": station_id,
            "station_name": station_name,
            "area_km2": area_km2,
            "file_name": fname,
            "reason": str(e)
        })


# =========================
# FINAL SUMMARY TABLES
# =========================

summary_df = pd.DataFrame(all_results)
failed_df = pd.DataFrame(failed_results)

if summary_df.empty:
    print("No validation results produced.")
else:
    summary_df = summary_df.sort_values("station_id").reset_index(drop=True)

    summary_path = os.path.join(
        output_folder,
        "Kirchner_2010_calibration_2011_2022_validation_summary_manual_station_list.csv"
    )

    summary_df.to_csv(summary_path, index=False)

    display(summary_df)

    print("\nSaved summary table:")
    print(summary_path)

if not failed_df.empty:
    failed_df = failed_df.sort_values("station_id").reset_index(drop=True)

    failed_path = os.path.join(
        output_folder,
        "failed_or_skipped_stations.csv"
    )

    failed_df.to_csv(failed_path, index=False)

    print("\nSome stations were skipped or failed. Saved list:")
    print(failed_path)

print("\nAll outputs saved to:")
print(output_folder)
```

```python
# ============================================================
# REVISED ANNUAL DISCHARGE AND DYNAMIC GROUNDWATER STORAGE
# Increased K for storage calculation only
# ============================================================
#
# IMPORTANT:
#
# 1. Observed discharge is not modified.
#
# 2. The recession analysis estimates:
#
#       K_fitted_days = 1 / a_linear
#
# 3. Dynamic storage is calculated using:
#
#       K_storage_days =
#           K_fitted_days * K_storage_multiplier
#
# 4. Storage:
#
#       S_mm = K_storage_days * Q_mm_day
#
# 5. Set:
#
#       K_storage_multiplier = 1.0
#
#    for the original fitted storage.
#
#       K_storage_multiplier = 2.0
#
#    doubles the storage.
#
#       K_storage_multiplier = 3.0
#
#    triples the storage.
#
# ============================================================

import os
import shutil
import calendar
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ============================================================
# 1. SETTINGS
# ============================================================

revised_output_folder = os.path.join(
    output_folder,
    "revised_annual_discharge_and_storage"
)

plot_output_folder = os.path.join(
    revised_output_folder,
    "plots"
)

station_output_folder = os.path.join(
    revised_output_folder,
    "station_csv"
)

os.makedirs(
    revised_output_folder,
    exist_ok=True
)

os.makedirs(
    plot_output_folder,
    exist_ok=True
)

os.makedirs(
    station_output_folder,
    exist_ok=True
)


# Minimum continuous recession length
linear_min_recession_length = 7


# ============================================================
# STORAGE K MULTIPLIER
# ============================================================
#
# This multiplier changes storage only.
# It does not change discharge.
#
# 1.0 = original fitted storage
# 1.5 = 50% larger storage
# 2.0 = double storage
# 3.0 = triple storage
#
K_storage_multiplier = 3.0


# Optional adjusted-K limits
# Leave as None if no limits are needed
minimum_K_storage_days = None
maximum_K_storage_days = None


# Minimum annual data coverage
minimum_annual_coverage_percent = 80.0


# Remove negative discharge values
remove_negative_discharge = True


# ============================================================
# 2. READ DAILY DISCHARGE AND CONVERT TO MM/DAY
# ============================================================

def prepare_daily_discharge_depth_revised(
    file_path,
    area_km2
):
    """
    Read discharge and convert m3/s into mm/day.

    Q_mm_day = Q_m3_s * 86.4 / area_km2
    """

    if pd.isna(area_km2) or area_km2 <= 0:
        raise ValueError(
            "Catchment area must be greater than zero."
        )

    df = read_table_flexible(file_path)

    df.columns = (
        df.columns
        .astype(str)
        .str.strip()
    )

    date_col = detect_date_column(df)
    q_col = detect_discharge_column(df)

    df[date_col] = pd.to_datetime(
        df[date_col],
        errors="coerce"
    )

    df[q_col] = pd.to_numeric(
        df[q_col],
        errors="coerce"
    )

    df = df[
        [date_col, q_col]
    ].dropna()

    if df.empty:
        return pd.DataFrame()

    df = (
        df
        .set_index(date_col)
        .sort_index()
    )

    df = df.loc[
        f"{start_year}-01-01":
        f"{end_year}-12-31"
    ]

    if df.empty:
        return pd.DataFrame()

    # Convert sub-daily data to daily mean
    df = df.resample("D").mean()

    df = df.rename(
        columns={
            q_col: "Q_m3_s"
        }
    )

    df["Q_m3_s"] = pd.to_numeric(
        df["Q_m3_s"],
        errors="coerce"
    )

    if remove_negative_discharge:
        df.loc[
            df["Q_m3_s"] < 0,
            "Q_m3_s"
        ] = np.nan

    # Convert m3/s to mm/day
    df["Q_mm_day"] = (
        df["Q_m3_s"]
        * 86.4
        / area_km2
    )

    df["year"] = df.index.year

    return df


# ============================================================
# 3. EXTRACT CONTINUOUS RECESSION EVENTS
# ============================================================

def prepare_linear_recession_data(
    daily_df,
    calibration_year,
    min_recession_length=7
):
    """
    Extract continuous declining-flow events.

    Linear recession equation:

        -dQ/dt = a_linear * Q
    """

    if daily_df.empty:
        return pd.DataFrame()

    calibration_df = daily_df[
        daily_df.index.year
        == calibration_year
    ].copy()

    calibration_df = (
        calibration_df
        .dropna(
            subset=["Q_mm_day"]
        )
    )

    if len(calibration_df) < 20:
        return pd.DataFrame()

    calibration_df["Q_next"] = (
        calibration_df[
            "Q_mm_day"
        ].shift(-1)
    )

    calibration_df["date_next"] = (
        calibration_df
        .index
        .to_series()
        .shift(-1)
    )

    calibration_df["day_difference"] = (
        calibration_df["date_next"]
        - calibration_df
        .index
        .to_series()
    ).dt.days

    calibration_df["declining_pair"] = (
        (
            calibration_df["Q_next"]
            < calibration_df["Q_mm_day"]
        )
        & (
            calibration_df["day_difference"]
            == 1
        )
        & (
            calibration_df["Q_mm_day"]
            > 0
        )
        & (
            calibration_df["Q_next"]
            > 0
        )
    )

    calibration_df["event_break"] = (
        ~calibration_df["declining_pair"]
    )

    calibration_df["recession_group"] = (
        calibration_df[
            "event_break"
        ].cumsum()
    )

    event_lengths = (
        calibration_df
        .groupby(
            "recession_group"
        )["declining_pair"]
        .transform("sum")
    )

    recession_df = calibration_df[
        calibration_df["declining_pair"]
        & (
            event_lengths
            >= min_recession_length
        )
    ].copy()

    if recession_df.empty:
        return pd.DataFrame()

    recession_df["Q_avg_mm_day"] = (
        recession_df["Q_mm_day"]
        + recession_df["Q_next"]
    ) / 2.0

    recession_df[
        "minus_dQdt_mm_day2"
    ] = -(
        recession_df["Q_next"]
        - recession_df["Q_mm_day"]
    )

    recession_df = (
        recession_df
        .replace(
            [np.inf, -np.inf],
            np.nan
        )
    )

    recession_df = (
        recession_df
        .dropna(
            subset=[
                "Q_avg_mm_day",
                "minus_dQdt_mm_day2"
            ]
        )
    )

    recession_df = recession_df[
        (
            recession_df[
                "Q_avg_mm_day"
            ] > 0
        )
        & (
            recession_df[
                "minus_dQdt_mm_day2"
            ] > 0
        )
    ].copy()

    return recession_df


# ============================================================
# 4. FIT LINEAR RECESSION COEFFICIENT
# ============================================================

def fit_linear_recession_coefficient(
    recession_df
):
    """
    Fit:

        -dQ/dt = a_linear * Q

    using least squares through the origin.

    Returns:
        a_linear
        K_fitted_days
        r_squared
    """

    if recession_df.empty:
        return np.nan, np.nan, np.nan

    x = recession_df[
        "Q_avg_mm_day"
    ].to_numpy(dtype=float)

    y = recession_df[
        "minus_dQdt_mm_day2"
    ].to_numpy(dtype=float)

    valid = (
        np.isfinite(x)
        & np.isfinite(y)
        & (x > 0)
        & (y > 0)
    )

    x = x[valid]
    y = y[valid]

    if len(x) < 10:
        return np.nan, np.nan, np.nan

    denominator = np.sum(
        x ** 2
    )

    if denominator <= 0:
        return np.nan, np.nan, np.nan

    a_linear = (
        np.sum(x * y)
        / denominator
    )

    if (
        not np.isfinite(a_linear)
        or a_linear <= 0
    ):
        return np.nan, np.nan, np.nan

    K_fitted_days = (
        1.0 / a_linear
    )

    predicted = (
        a_linear * x
    )

    ss_res = np.sum(
        (y - predicted) ** 2
    )

    ss_tot = np.sum(
        (y - np.mean(y)) ** 2
    )

    if ss_tot > 0:
        r_squared = (
            1.0
            - ss_res / ss_tot
        )
    else:
        r_squared = np.nan

    return (
        a_linear,
        K_fitted_days,
        r_squared
    )


# ============================================================
# 5. EXPECTED NUMBER OF DAYS
# ============================================================

def expected_days_in_year(year):
    """
    Return 365 or 366 days.
    """

    return (
        366
        if calendar.isleap(int(year))
        else 365
    )


# ============================================================
# 6. EMPTY RESULT CONTAINERS
# ============================================================

annual_results = []
station_summary_results = []
failed_station_results = []
linear_recession_results = []


# ============================================================
# 7. PROCESS EVERY STATION
# ============================================================

for _, station_row in run_df.iterrows():

    station_id = int(
        station_row["station_id"]
    )

    station_name = station_row.get(
        "station_name",
        ""
    )

    area_km2 = station_row.get(
        "area_km2",
        np.nan
    )

    file_path = station_row["file"]

    print(
        "\nProcessing:",
        station_id,
        "|",
        station_name
    )

    try:

        if (
            pd.isna(area_km2)
            or area_km2 <= 0
        ):
            raise ValueError(
                "Missing or invalid catchment area."
            )

        # ----------------------------------------------------
        # Read daily discharge
        # ----------------------------------------------------

        daily_df = (
            prepare_daily_discharge_depth_revised(
                file_path=file_path,
                area_km2=area_km2
            )
        )

        if daily_df.empty:
            raise ValueError(
                "No valid daily discharge data."
            )

        # ----------------------------------------------------
        # Prepare recession observations
        # ----------------------------------------------------

        recession_df = (
            prepare_linear_recession_data(
                daily_df=daily_df,
                calibration_year=calibration_year,
                min_recession_length=(
                    linear_min_recession_length
                )
            )
        )

        # ----------------------------------------------------
        # Estimate original fitted K
        # ----------------------------------------------------

        (
            a_linear,
            K_fitted_days,
            recession_r_squared
        ) = fit_linear_recession_coefficient(
            recession_df
        )

        if not np.isfinite(
            K_fitted_days
        ):
            raise ValueError(
                "A valid K could not be estimated."
            )

        # ----------------------------------------------------
        # Increase K for storage only
        # ----------------------------------------------------

        K_storage_days = (
            K_fitted_days
            * K_storage_multiplier
        )

        if (
            minimum_K_storage_days
            is not None
        ):
            K_storage_days = max(
                K_storage_days,
                float(
                    minimum_K_storage_days
                )
            )

        if (
            maximum_K_storage_days
            is not None
        ):
            K_storage_days = min(
                K_storage_days,
                float(
                    maximum_K_storage_days
                )
            )

        print(
            f"a = {a_linear:.6f} 1/day | "
            f"Fitted K = "
            f"{K_fitted_days:.2f} days | "
            f"Storage K = "
            f"{K_storage_days:.2f} days | "
            f"Multiplier = "
            f"{K_storage_multiplier:.2f} | "
            f"R² = "
            f"{recession_r_squared:.3f}"
        )

        linear_recession_results.append({
            "station_id":
                station_id,

            "station_name":
                station_name,

            "area_km2":
                area_km2,

            "calibration_year":
                calibration_year,

            "a_linear_per_day":
                a_linear,

            "K_fitted_days":
                K_fitted_days,

            "K_storage_multiplier":
                K_storage_multiplier,

            "K_storage_days":
                K_storage_days,

            "recession_R2":
                recession_r_squared,

            "n_recession_points":
                len(recession_df),

            "min_recession_length_days":
                linear_min_recession_length
        })

        # ----------------------------------------------------
        # Dynamic storage using increased K
        # ----------------------------------------------------

        daily_df["storage_mm"] = (
            K_storage_days
            * daily_df["Q_mm_day"]
        )

        daily_df["storage_Mm3"] = (
            daily_df["storage_mm"]
            * area_km2
            / 1000.0
        )

        # ----------------------------------------------------
        # Seven-day discharge average
        # ----------------------------------------------------

        daily_df["Q7_mm_day"] = (
            daily_df
            .groupby("year")[
                "Q_mm_day"
            ]
            .transform(
                lambda series:
                series.rolling(
                    window=7,
                    min_periods=7
                ).mean()
            )
        )

        # Increased-K low-flow storage
        daily_df["storage_Q7_mm"] = (
            K_storage_days
            * daily_df["Q7_mm_day"]
        )

        daily_df[
            "storage_Q7_Mm3"
        ] = (
            daily_df[
                "storage_Q7_mm"
            ]
            * area_km2
            / 1000.0
        )

        # ----------------------------------------------------
        # Annual statistics
        # ----------------------------------------------------

        annual_station_df = (
            daily_df
            .groupby("year")
            .agg(
                days_available=(
                    "Q_mm_day",
                    "count"
                ),

                mean_discharge_m3_s=(
                    "Q_m3_s",
                    "mean"
                ),

                mean_discharge_mm_day=(
                    "Q_mm_day",
                    "mean"
                ),

                annual_discharge_mm_year=(
                    "Q_mm_day",
                    "sum"
                ),

                mean_storage_mm=(
                    "storage_mm",
                    "mean"
                ),

                median_storage_mm=(
                    "storage_mm",
                    "median"
                ),

                minimum_storage_mm=(
                    "storage_mm",
                    "min"
                ),

                maximum_storage_mm=(
                    "storage_mm",
                    "max"
                ),

                mean_storage_Mm3=(
                    "storage_Mm3",
                    "mean"
                ),

                median_storage_Mm3=(
                    "storage_Mm3",
                    "median"
                ),

                minimum_storage_Mm3=(
                    "storage_Mm3",
                    "min"
                ),

                maximum_storage_Mm3=(
                    "storage_Mm3",
                    "max"
                ),

                minimum_7day_discharge_mm_day=(
                    "Q7_mm_day",
                    "min"
                ),

                lowflow_storage_mm=(
                    "storage_Q7_mm",
                    "min"
                ),

                lowflow_storage_Mm3=(
                    "storage_Q7_Mm3",
                    "min"
                )
            )
            .reset_index()
        )

        # ----------------------------------------------------
        # Annual discharge volume
        # ----------------------------------------------------

        annual_station_df[
            "annual_discharge_Mm3_year"
        ] = (
            annual_station_df[
                "annual_discharge_mm_year"
            ]
            * area_km2
            / 1000.0
        )

        # ----------------------------------------------------
        # Annual coverage
        # ----------------------------------------------------

        annual_station_df[
            "expected_days"
        ] = (
            annual_station_df[
                "year"
            ].apply(
                expected_days_in_year
            )
        )

        annual_station_df[
            "coverage_percent"
        ] = (
            annual_station_df[
                "days_available"
            ]
            / annual_station_df[
                "expected_days"
            ]
            * 100.0
        )

        annual_station_df[
            "coverage_flag"
        ] = np.where(
            annual_station_df[
                "coverage_percent"
            ]
            >= minimum_annual_coverage_percent,
            "Acceptable",
            "Incomplete year"
        )

        # ----------------------------------------------------
        # Add station and K information
        # ----------------------------------------------------

        annual_station_df[
            "station_id"
        ] = station_id

        annual_station_df[
            "station_name"
        ] = station_name

        annual_station_df[
            "area_km2"
        ] = area_km2

        annual_station_df[
            "a_linear_per_day"
        ] = a_linear

        annual_station_df[
            "K_fitted_days"
        ] = K_fitted_days

        annual_station_df[
            "K_storage_multiplier"
        ] = K_storage_multiplier

        annual_station_df[
            "K_storage_days"
        ] = K_storage_days

        annual_station_df[
            "recession_R2"
        ] = recession_r_squared

        # ----------------------------------------------------
        # Reorder columns
        # ----------------------------------------------------

        annual_station_df = annual_station_df[
            [
                "station_id",
                "station_name",
                "area_km2",
                "year",
                "days_available",
                "expected_days",
                "coverage_percent",
                "coverage_flag",
                "a_linear_per_day",
                "K_fitted_days",
                "K_storage_multiplier",
                "K_storage_days",
                "recession_R2",
                "mean_discharge_m3_s",
                "mean_discharge_mm_day",
                "annual_discharge_mm_year",
                "annual_discharge_Mm3_year",
                "mean_storage_mm",
                "median_storage_mm",
                "minimum_storage_mm",
                "maximum_storage_mm",
                "mean_storage_Mm3",
                "median_storage_Mm3",
                "minimum_storage_Mm3",
                "maximum_storage_Mm3",
                "minimum_7day_discharge_mm_day",
                "lowflow_storage_mm",
                "lowflow_storage_Mm3"
            ]
        ]

        annual_results.append(
            annual_station_df
        )

        # ----------------------------------------------------
        # Save station annual CSV
        # ----------------------------------------------------

        station_csv_path = os.path.join(
            station_output_folder,
            (
                f"annual_results_station_"
                f"{station_id}.csv"
            )
        )

        annual_station_df.to_csv(
            station_csv_path,
            index=False
        )

        # ----------------------------------------------------
        # Save recession CSV
        # ----------------------------------------------------

        recession_output = (
            recession_df.copy()
        )

        recession_output[
            "station_id"
        ] = station_id

        recession_output[
            "station_name"
        ] = station_name

        recession_output[
            "area_km2"
        ] = area_km2

        recession_output[
            "a_linear_per_day"
        ] = a_linear

        recession_output[
            "K_fitted_days"
        ] = K_fitted_days

        recession_output[
            "K_storage_multiplier"
        ] = K_storage_multiplier

        recession_output[
            "K_storage_days"
        ] = K_storage_days

        recession_csv_path = os.path.join(
            station_output_folder,
            (
                f"linear_recession_station_"
                f"{station_id}.csv"
            )
        )

        recession_output.to_csv(
            recession_csv_path,
            index=True
        )

        # ----------------------------------------------------
        # Select acceptable years
        # ----------------------------------------------------

        acceptable_years = annual_station_df[
            annual_station_df[
                "coverage_flag"
            ]
            == "Acceptable"
        ].copy()

        if acceptable_years.empty:
            summary_source = (
                annual_station_df
            )
        else:
            summary_source = (
                acceptable_years
            )

        # ----------------------------------------------------
        # Station summary
        # ----------------------------------------------------

        station_summary_results.append({
            "station_id":
                station_id,

            "station_name":
                station_name,

            "area_km2":
                area_km2,

            "a_linear_per_day":
                a_linear,

            "K_fitted_days":
                K_fitted_days,

            "K_storage_multiplier":
                K_storage_multiplier,

            "K_storage_days":
                K_storage_days,

            "recession_R2":
                recession_r_squared,

            "n_recession_points":
                len(recession_df),

            "n_years_total":
                len(annual_station_df),

            "n_acceptable_years":
                len(acceptable_years),

            "mean_annual_discharge_mm_year":
                summary_source[
                    "annual_discharge_mm_year"
                ].mean(),

            "median_annual_discharge_mm_year":
                summary_source[
                    "annual_discharge_mm_year"
                ].median(),

            "minimum_annual_discharge_mm_year":
                summary_source[
                    "annual_discharge_mm_year"
                ].min(),

            "maximum_annual_discharge_mm_year":
                summary_source[
                    "annual_discharge_mm_year"
                ].max(),

            "mean_annual_discharge_Mm3_year":
                summary_source[
                    "annual_discharge_Mm3_year"
                ].mean(),

            "median_annual_discharge_Mm3_year":
                summary_source[
                    "annual_discharge_Mm3_year"
                ].median(),

            "minimum_annual_discharge_Mm3_year":
                summary_source[
                    "annual_discharge_Mm3_year"
                ].min(),

            "maximum_annual_discharge_Mm3_year":
                summary_source[
                    "annual_discharge_Mm3_year"
                ].max(),

            "mean_dynamic_storage_mm":
                summary_source[
                    "mean_storage_mm"
                ].mean(),

            "median_dynamic_storage_mm":
                summary_source[
                    "median_storage_mm"
                ].median(),

            "minimum_dynamic_storage_mm":
                summary_source[
                    "minimum_storage_mm"
                ].min(),

            "maximum_dynamic_storage_mm":
                summary_source[
                    "maximum_storage_mm"
                ].max(),

            "mean_dynamic_storage_Mm3":
                summary_source[
                    "mean_storage_Mm3"
                ].mean(),

            "median_dynamic_storage_Mm3":
                summary_source[
                    "median_storage_Mm3"
                ].median(),

            "minimum_dynamic_storage_Mm3":
                summary_source[
                    "minimum_storage_Mm3"
                ].min(),

            "maximum_dynamic_storage_Mm3":
                summary_source[
                    "maximum_storage_Mm3"
                ].max(),

            "mean_minimum_7day_discharge_mm_day":
                summary_source[
                    "minimum_7day_discharge_mm_day"
                ].mean(),

            "mean_lowflow_storage_mm":
                summary_source[
                    "lowflow_storage_mm"
                ].mean(),

            "median_lowflow_storage_mm":
                summary_source[
                    "lowflow_storage_mm"
                ].median(),

            "mean_lowflow_storage_Mm3":
                summary_source[
                    "lowflow_storage_Mm3"
                ].mean(),

            "median_lowflow_storage_Mm3":
                summary_source[
                    "lowflow_storage_Mm3"
                ].median()
        })

        # ====================================================
        # 8. STATION PLOTS
        # ====================================================

        # Annual discharge depth
        plt.figure(figsize=(9, 5))

        plt.plot(
            annual_station_df["year"],
            annual_station_df[
                "annual_discharge_mm_year"
            ],
            marker="o",
            linewidth=2
        )

        plt.xlabel("Year")

        plt.ylabel(
            "Annual discharge depth (mm/year)"
        )

        plt.title(
            f"Annual Discharge Depth\n"
            f"Station {station_id} – "
            f"{station_name}"
        )

        plt.grid(
            True,
            linestyle="--",
            alpha=0.4
        )

        plt.tight_layout()

        plt.savefig(
            os.path.join(
                plot_output_folder,
                (
                    f"annual_discharge_mm_"
                    f"station_{station_id}.png"
                )
            ),
            dpi=300,
            bbox_inches="tight"
        )

        plt.close()

        # Annual discharge volume
        plt.figure(figsize=(9, 5))

        plt.plot(
            annual_station_df["year"],
            annual_station_df[
                "annual_discharge_Mm3_year"
            ],
            marker="o",
            linewidth=2
        )

        plt.xlabel("Year")

        plt.ylabel(
            "Annual discharge volume "
            "(Mm³/year)"
        )

        plt.title(
            f"Annual Discharge Volume\n"
            f"Station {station_id} – "
            f"{station_name}"
        )

        plt.grid(
            True,
            linestyle="--",
            alpha=0.4
        )

        plt.tight_layout()

        plt.savefig(
            os.path.join(
                plot_output_folder,
                (
                    f"annual_discharge_Mm3_"
                    f"station_{station_id}.png"
                )
            ),
            dpi=300,
            bbox_inches="tight"
        )

        plt.close()

        # Dynamic storage depth
        plt.figure(figsize=(9, 5))

        plt.plot(
            annual_station_df["year"],
            annual_station_df[
                "mean_storage_mm"
            ],
            marker="o",
            linewidth=2,
            label="Annual mean storage"
        )

        plt.plot(
            annual_station_df["year"],
            annual_station_df[
                "lowflow_storage_mm"
            ],
            marker="s",
            linewidth=1.5,
            label="Annual low-flow storage"
        )

        plt.xlabel("Year")

        plt.ylabel(
            "Dynamic groundwater storage "
            "depth (mm)"
        )

        plt.title(
            f"Dynamic Groundwater Storage\n"
            f"Station {station_id} – "
            f"{station_name}\n"
            f"Fitted K = "
            f"{K_fitted_days:.1f} days, "
            f"Storage K = "
            f"{K_storage_days:.1f} days"
        )

        plt.grid(
            True,
            linestyle="--",
            alpha=0.4
        )

        plt.legend()
        plt.tight_layout()

        plt.savefig(
            os.path.join(
                plot_output_folder,
                (
                    f"dynamic_storage_mm_"
                    f"station_{station_id}.png"
                )
            ),
            dpi=300,
            bbox_inches="tight"
        )

        plt.close()

        # Dynamic storage volume
        plt.figure(figsize=(9, 5))

        plt.plot(
            annual_station_df["year"],
            annual_station_df[
                "mean_storage_Mm3"
            ],
            marker="o",
            linewidth=2,
            label="Annual mean storage"
        )

        plt.plot(
            annual_station_df["year"],
            annual_station_df[
                "lowflow_storage_Mm3"
            ],
            marker="s",
            linewidth=1.5,
            label="Annual low-flow storage"
        )

        plt.xlabel("Year")

        plt.ylabel(
            "Dynamic groundwater storage "
            "volume (Mm³)"
        )

        plt.title(
            f"Dynamic Groundwater Storage Volume\n"
            f"Station {station_id} – "
            f"{station_name}\n"
            f"Storage K multiplier = "
            f"{K_storage_multiplier:.1f}"
        )

        plt.grid(
            True,
            linestyle="--",
            alpha=0.4
        )

        plt.legend()
        plt.tight_layout()

        plt.savefig(
            os.path.join(
                plot_output_folder,
                (
                    f"dynamic_storage_Mm3_"
                    f"station_{station_id}.png"
                )
            ),
            dpi=300,
            bbox_inches="tight"
        )

        plt.close()

        # Recession-fit plot
        plt.figure(figsize=(8, 6))

        plt.scatter(
            recession_df[
                "Q_avg_mm_day"
            ],
            recession_df[
                "minus_dQdt_mm_day2"
            ],
            s=12,
            alpha=0.4,
            label="Observed recession points"
        )

        q_plot = np.linspace(
            recession_df[
                "Q_avg_mm_day"
            ].min(),

            recession_df[
                "Q_avg_mm_day"
            ].max(),

            200
        )

        plt.plot(
            q_plot,
            a_linear * q_plot,
            linewidth=2.5,
            label=(
                f"Linear fit: -dQ/dt = "
                f"{a_linear:.4f}Q"
            )
        )

        plt.xlabel(
            "Discharge depth, Q (mm/day)"
        )

        plt.ylabel(
            "Recession rate, −dQ/dt "
            "(mm/day²)"
        )

        plt.title(
            f"Linear Recession Fit – "
            f"{calibration_year}\n"
            f"Station {station_id} – "
            f"{station_name}\n"
            f"Fitted K = "
            f"{K_fitted_days:.1f} days, "
            f"Storage K = "
            f"{K_storage_days:.1f} days, "
            f"R² = "
            f"{recession_r_squared:.2f}"
        )

        plt.grid(
            True,
            linestyle="--",
            alpha=0.4
        )

        plt.legend()
        plt.tight_layout()

        plt.savefig(
            os.path.join(
                plot_output_folder,
                (
                    f"linear_recession_fit_"
                    f"station_{station_id}.png"
                )
            ),
            dpi=300,
            bbox_inches="tight"
        )

        plt.close()

        print(
            f"Completed station {station_id}: "
            f"mean annual discharge = "
            f"{annual_station_df['annual_discharge_mm_year'].mean():.2f} "
            f"mm/year; "
            f"mean dynamic storage = "
            f"{annual_station_df['mean_storage_mm'].mean():.2f} "
            f"mm"
        )

    except Exception as error:

        print(
            f"Failed station "
            f"{station_id}: {error}"
        )

        failed_station_results.append({
            "station_id":
                station_id,

            "station_name":
                station_name,

            "area_km2":
                area_km2,

            "file":
                file_path,

            "reason":
                str(error)
        })


# ============================================================
# 9. COMBINE ANNUAL RESULTS
# ============================================================

if len(annual_results) == 0:

    annual_results_df = pd.DataFrame()

    print(
        "\nNo annual results were produced."
    )

else:

    annual_results_df = pd.concat(
        annual_results,
        ignore_index=True
    )

    annual_results_df = (
        annual_results_df
        .sort_values(
            ["station_id", "year"]
        )
        .reset_index(drop=True)
    )

    annual_results_csv = os.path.join(
        revised_output_folder,
        (
            "annual_discharge_and_"
            "dynamic_storage_2010_2022.csv"
        )
    )

    annual_results_df.to_csv(
        annual_results_csv,
        index=False
    )

    print(
        "\nCombined annual results saved:"
    )

    print(annual_results_csv)

    display(
        annual_results_df.head()
    )


# ============================================================
# 10. STATION SUMMARY
# ============================================================

station_summary_df = pd.DataFrame(
    station_summary_results
)

if not station_summary_df.empty:

    station_summary_df = (
        station_summary_df
        .sort_values("station_id")
        .reset_index(drop=True)
    )

    station_summary_csv = os.path.join(
        revised_output_folder,
        (
            "station_summary_discharge_"
            "and_storage_2010_2022.csv"
        )
    )

    station_summary_df.to_csv(
        station_summary_csv,
        index=False
    )

    print(
        "\nStation summary saved:"
    )

    print(station_summary_csv)

    display(
        station_summary_df.head()
    )


# ============================================================
# 11. RECESSION PARAMETERS
# ============================================================

linear_recession_df = pd.DataFrame(
    linear_recession_results
)

if not linear_recession_df.empty:

    linear_recession_df = (
        linear_recession_df
        .sort_values("station_id")
        .reset_index(drop=True)
    )

    recession_parameter_csv = os.path.join(
        revised_output_folder,
        (
            "linear_recession_parameters_"
            f"{calibration_year}.csv"
        )
    )

    linear_recession_df.to_csv(
        recession_parameter_csv,
        index=False
    )


# ============================================================
# 12. FAILED STATIONS
# ============================================================

failed_station_df = pd.DataFrame(
    failed_station_results
)

if not failed_station_df.empty:

    failed_station_csv = os.path.join(
        revised_output_folder,
        "failed_stations.csv"
    )

    failed_station_df.to_csv(
        failed_station_csv,
        index=False
    )

    print(
        "\nSome stations failed:"
    )

    print(failed_station_csv)


# ============================================================
# 13. COMPARISON BAR PLOTS
# ============================================================

if not station_summary_df.empty:

    # Mean annual discharge depth
    plot_df = (
        station_summary_df
        .sort_values(
            "mean_annual_discharge_mm_year",
            ascending=False
        )
    )

    plt.figure(
        figsize=(
            max(
                12,
                len(plot_df) * 0.35
            ),
            7
        )
    )

    plt.bar(
        plot_df[
            "station_id"
        ].astype(str),

        plot_df[
            "mean_annual_discharge_mm_year"
        ]
    )

    plt.xlabel("Station ID")

    plt.ylabel(
        "Mean annual discharge depth "
        "(mm/year)"
    )

    plt.title(
        "Mean Annual Discharge Depth "
        "by Station, 2010–2022"
    )

    plt.xticks(rotation=90)

    plt.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            plot_output_folder,
            (
                "barplot_mean_annual_"
                "discharge_mm_year.png"
            )
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

    # Mean annual discharge volume
    plot_df = (
        station_summary_df
        .sort_values(
            "mean_annual_discharge_Mm3_year",
            ascending=False
        )
    )

    plt.figure(
        figsize=(
            max(
                12,
                len(plot_df) * 0.35
            ),
            7
        )
    )

    plt.bar(
        plot_df[
            "station_id"
        ].astype(str),

        plot_df[
            "mean_annual_discharge_Mm3_year"
        ]
    )

    plt.xlabel("Station ID")

    plt.ylabel(
        "Mean annual discharge volume "
        "(Mm³/year)"
    )

    plt.title(
        "Mean Annual Discharge Volume "
        "by Station, 2010–2022"
    )

    plt.xticks(rotation=90)

    plt.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            plot_output_folder,
            (
                "barplot_mean_annual_"
                "discharge_Mm3_year.png"
            )
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

    # Mean dynamic storage depth
    plot_df = (
        station_summary_df
        .sort_values(
            "mean_dynamic_storage_mm",
            ascending=False
        )
    )

    plt.figure(
        figsize=(
            max(
                12,
                len(plot_df) * 0.35
            ),
            7
        )
    )

    plt.bar(
        plot_df[
            "station_id"
        ].astype(str),

        plot_df[
            "mean_dynamic_storage_mm"
        ]
    )

    plt.xlabel("Station ID")

    plt.ylabel(
        "Mean dynamic storage depth (mm)"
    )

    plt.title(
        "Mean Dynamic Groundwater Storage "
        f"by Station, 2010–2022\n"
        f"K multiplier = "
        f"{K_storage_multiplier:.1f}"
    )

    plt.xticks(rotation=90)

    plt.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            plot_output_folder,
            (
                "barplot_mean_dynamic_"
                "storage_mm.png"
            )
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

    # Mean dynamic storage volume
    plot_df = (
        station_summary_df
        .sort_values(
            "mean_dynamic_storage_Mm3",
            ascending=False
        )
    )

    plt.figure(
        figsize=(
            max(
                12,
                len(plot_df) * 0.35
            ),
            7
        )
    )

    plt.bar(
        plot_df[
            "station_id"
        ].astype(str),

        plot_df[
            "mean_dynamic_storage_Mm3"
        ]
    )

    plt.xlabel("Station ID")

    plt.ylabel(
        "Mean dynamic storage volume "
        "(Mm³)"
    )

    plt.title(
        "Mean Dynamic Groundwater Storage "
        "Volume by Station, 2010–2022"
    )

    plt.xticks(rotation=90)

    plt.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            plot_output_folder,
            (
                "barplot_mean_dynamic_"
                "storage_Mm3.png"
            )
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

    # Low-flow storage depth
    plot_df = (
        station_summary_df
        .sort_values(
            "mean_lowflow_storage_mm",
            ascending=False
        )
    )

    plt.figure(
        figsize=(
            max(
                12,
                len(plot_df) * 0.35
            ),
            7
        )
    )

    plt.bar(
        plot_df[
            "station_id"
        ].astype(str),

        plot_df[
            "mean_lowflow_storage_mm"
        ]
    )

    plt.xlabel("Station ID")

    plt.ylabel(
        "Mean low-flow storage depth (mm)"
    )

    plt.title(
        "Mean Seven-Day Low-Flow Storage "
        "by Station, 2010–2022"
    )

    plt.xticks(rotation=90)

    plt.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            plot_output_folder,
            (
                "barplot_mean_lowflow_"
                "storage_mm.png"
            )
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()


# ============================================================
# 14. HEATMAPS
# ============================================================

if not annual_results_df.empty:

    # Annual discharge heatmap
    discharge_heatmap = (
        annual_results_df
        .pivot_table(
            index="station_id",
            columns="year",
            values=(
                "annual_discharge_mm_year"
            ),
            aggfunc="mean"
        )
        .sort_index()
    )

    plt.figure(
        figsize=(
            13,
            max(
                6,
                0.28
                * len(discharge_heatmap)
            )
        )
    )

    image = plt.imshow(
        discharge_heatmap.values,
        aspect="auto"
    )

    plt.colorbar(
        image,
        label=(
            "Annual discharge "
            "(mm/year)"
        )
    )

    plt.xticks(
        np.arange(
            len(
                discharge_heatmap.columns
            )
        ),
        discharge_heatmap.columns,
        rotation=45
    )

    plt.yticks(
        np.arange(
            len(
                discharge_heatmap.index
            )
        ),
        discharge_heatmap.index
    )

    plt.xlabel("Year")
    plt.ylabel("Station ID")

    plt.title(
        "Annual Discharge Depth, "
        "2010–2022"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            plot_output_folder,
            (
                "heatmap_annual_discharge_"
                "mm_year.png"
            )
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

    # Mean storage heatmap
    storage_heatmap = (
        annual_results_df
        .pivot_table(
            index="station_id",
            columns="year",
            values="mean_storage_mm",
            aggfunc="mean"
        )
        .sort_index()
    )

    plt.figure(
        figsize=(
            13,
            max(
                6,
                0.28
                * len(storage_heatmap)
            )
        )
    )

    image = plt.imshow(
        storage_heatmap.values,
        aspect="auto"
    )

    plt.colorbar(
        image,
        label=(
            "Mean dynamic storage "
            "depth (mm)"
        )
    )

    plt.xticks(
        np.arange(
            len(
                storage_heatmap.columns
            )
        ),
        storage_heatmap.columns,
        rotation=45
    )

    plt.yticks(
        np.arange(
            len(
                storage_heatmap.index
            )
        ),
        storage_heatmap.index
    )

    plt.xlabel("Year")
    plt.ylabel("Station ID")

    plt.title(
        "Annual Mean Dynamic Groundwater "
        f"Storage, 2010–2022\n"
        f"K multiplier = "
        f"{K_storage_multiplier:.1f}"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            plot_output_folder,
            (
                "heatmap_mean_dynamic_"
                "storage_mm.png"
            )
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

    # Low-flow storage heatmap
    lowflow_heatmap = (
        annual_results_df
        .pivot_table(
            index="station_id",
            columns="year",
            values="lowflow_storage_mm",
            aggfunc="mean"
        )
        .sort_index()
    )

    plt.figure(
        figsize=(
            13,
            max(
                6,
                0.28
                * len(lowflow_heatmap)
            )
        )
    )

    image = plt.imshow(
        lowflow_heatmap.values,
        aspect="auto"
    )

    plt.colorbar(
        image,
        label=(
            "Low-flow dynamic storage "
            "depth (mm)"
        )
    )

    plt.xticks(
        np.arange(
            len(
                lowflow_heatmap.columns
            )
        ),
        lowflow_heatmap.columns,
        rotation=45
    )

    plt.yticks(
        np.arange(
            len(
                lowflow_heatmap.index
            )
        ),
        lowflow_heatmap.index
    )

    plt.xlabel("Year")
    plt.ylabel("Station ID")

    plt.title(
        "Seven-Day Low-Flow Dynamic "
        "Storage, 2010–2022"
    )

    plt.tight_layout()

    plt.savefig(
        os.path.join(
            plot_output_folder,
            (
                "heatmap_lowflow_"
                "storage_mm.png"
            )
        ),
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()


# ============================================================
# 15. SAVE EXCEL WORKBOOK
# ============================================================

excel_output_path = os.path.join(
    revised_output_folder,
    (
        "revised_discharge_and_storage_"
        "results_2010_2022.xlsx"
    )
)

try:

    with pd.ExcelWriter(
        excel_output_path,
        engine="openpyxl"
    ) as writer:

        if not annual_results_df.empty:

            annual_results_df.to_excel(
                writer,
                sheet_name="Annual_Results",
                index=False
            )

        if not station_summary_df.empty:

            station_summary_df.to_excel(
                writer,
                sheet_name="Station_Summary",
                index=False
            )

        if not linear_recession_df.empty:

            linear_recession_df.to_excel(
                writer,
                sheet_name=(
                    "Recession_Parameters"
                ),
                index=False
            )

        if not failed_station_df.empty:

            failed_station_df.to_excel(
                writer,
                sheet_name="Failed_Stations",
                index=False
            )

    print(
        "\nExcel workbook saved:"
    )

    print(excel_output_path)

except Exception as error:

    print(
        "\nExcel workbook could not be saved."
    )

    print(
        "Reason:",
        error
    )


# ============================================================
# 16. CREATE README FILE
# ============================================================

readme_path = os.path.join(
    revised_output_folder,
    "README.txt"
)

with open(
    readme_path,
    "w",
    encoding="utf-8"
) as file:

    file.write(
        "REVISED DISCHARGE AND DYNAMIC STORAGE ANALYSIS\n"
    )

    file.write(
        "================================================\n\n"
    )

    file.write(
        f"Study period: "
        f"{start_year}-{end_year}\n"
    )

    file.write(
        f"Calibration year: "
        f"{calibration_year}\n"
    )

    file.write(
        f"K storage multiplier: "
        f"{K_storage_multiplier}\n\n"
    )

    file.write(
        "K_fitted_days = 1 / a_linear\n"
    )

    file.write(
        "K_storage_days = "
        "K_fitted_days * "
        "K_storage_multiplier\n"
    )

    file.write(
        "Storage_mm = "
        "K_storage_days * Q_mm_day\n\n"
    )

    file.write(
        "Observed discharge was not modified.\n"
    )

    file.write(
        "Only dynamic-storage estimates use "
        "the adjusted K_storage_days.\n"
    )

    file.write(
        "Storage is reported in mm and Mm3, "
        "not mm/year or Mm3/year.\n"
    )


# ============================================================
# 17. CREATE ZIP FILE
# ============================================================

zip_base_path = os.path.join(
    output_folder,
    (
        "revised_discharge_and_storage_"
        "results_2010_2022"
    )
)

zip_file_path = shutil.make_archive(
    base_name=zip_base_path,
    format="zip",
    root_dir=revised_output_folder
)


# ============================================================
# 18. FINAL OUTPUT
# ============================================================

print(
    "\n================================================"
)

print(
    "REVISED ANALYSIS COMPLETED"
)

print(
    "================================================"
)

print(
    f"Storage K multiplier: "
    f"{K_storage_multiplier:.2f}"
)

print(
    "Observed discharge was not changed."
)

print(
    "Dynamic storage was calculated using "
    "the adjusted K_storage_days."
)

print(
    "\nOutput folder:"
)

print(
    revised_output_folder
)

print(
    "\nExcel file:"
)

print(
    excel_output_path
)

print(
    "\nZIP file:"
)

print(
    zip_file_path
)

print(
    "================================================"
)
```
